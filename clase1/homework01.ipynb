## Ejercicio 1 · Clasificar una edad (decisión)

Implementa `clasificar(edad)` que devuelva:
- `"menor"` si `edad < 18`
- `"adulto"` si `18 <= edad < 65`
- `"adulto mayor"` si `edad >= 65`

**Casos de ejemplo:** `clasificar(10)` → `"menor"`, `clasificar(40)` → `"adulto"`, `clasificar(70)` → `"adulto mayor"`.

FUNCIÓN clasificar(edad):

    SI edad < 18:
        RETORNAR "menor"
    SINO SI edad >= 18 Y edad < 65:
        RETORNAR "adulto"
    SINO:
        RETORNAR "adulto mayor"

```python
def clasificar(edad):
    if edad < 18:
      return "menor"
    elif 18 <= edad < 65:
      return "adulto"
    else:
      return "adulto mayor"
```

```python
# === Tests visibles · Ejercicio 1 ===
assert clasificar(10) == 'menor'
assert clasificar(40) == 'adulto'
assert clasificar(70) == 'adulto mayor'
print("✅ Ejercicio 1: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 1 ===
assert clasificar(17) == 'menor'
assert clasificar(18) == 'adulto'
assert clasificar(64) == 'adulto'
assert clasificar(65) == 'adulto mayor'
print("✅ Ejercicio 1: tests adicionales superados.")
```

## Ejercicio 2 · Suma de los pares hasta n (bucle + condición)

Implementa `suma_pares(n)` que devuelva la suma de todos los números **pares** desde 1 hasta `n` (inclusive).

**Casos de ejemplo:** `suma_pares(6)` → `12` (2+4+6); `suma_pares(1)` → `0`.

FUNCIÓN suma_pares(n):

    lista = rango de 0 hasta n (inclusive)
    suma = 0

    PARA cada numero EN lista:
        SI numero es divisible por 2 (numero % 2 == 0):
            suma = suma + numero

    RETORNAR suma

```python
def suma_pares(n):
  lista = range(0 , n + 1)
  suma = 0
  for n in lista:
    if n%2 == 0:
      suma = suma + n
  return suma
```

```python
# === Tests visibles · Ejercicio 2 ===
assert suma_pares(6) == 12
assert suma_pares(1) == 0
print("✅ Ejercicio 2: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 2 ===
assert suma_pares(2) == 2
assert suma_pares(10) == 30
assert suma_pares(0) == 0
print("✅ Ejercicio 2: tests adicionales superados.")
```

## Ejercicio 3 · Factorial (acumulador multiplicativo)

Implementa `factorial(n)` que devuelva `n!` = 1·2·3·…·n. Por convención, `factorial(0)` → `1`.

**Casos de ejemplo:** `factorial(5)` → `120`; `factorial(0)` → `1`.

FUNCIÓN factorial(n):

    lista = rango de 1 hasta n (inclusive)
    factorial = 1

    PARA cada numero EN lista:
        factorial = factorial * numero

    RETORNAR factorial

```python
def factorial(n):
    lista = range(1 , n + 1)
    factorial = 1
    for n in lista:
      factorial = factorial * n
    return factorial
```

```python
# === Tests visibles · Ejercicio 3 ===
assert factorial(5) == 120
assert factorial(0) == 1
print("✅ Ejercicio 3: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 3 ===
assert factorial(1) == 1
assert factorial(3) == 6
assert factorial(7) == 5040
print("✅ Ejercicio 3: tests adicionales superados.")
```

## Ejercicio 4 · Contar vocales (recorrer una cadena)

Implementa `contar_vocales(texto)` que cuente las vocales (`a, e, i, o, u`) en un texto **en minúsculas**.

**Casos de ejemplo:** `contar_vocales("murcielago")` → `5`; `contar_vocales("xyz")` → `0`.

FUNCIÓN contar_vocales(texto):

    contador = 0
    vocales = ("a", "e", "i", "o", "u")

    PARA cada letra EN texto:
        SI letra está en vocales:
            contador = contador + 1

    RETORNAR contador

```python
def contar_vocales(texto):
    contador = 0
    vocales = ("a", "e", "i", "o", "u")
    for l in texto:
        if l in vocales:
            contador += 1
    return contador
```

```python
# === Tests visibles · Ejercicio 4 ===
assert contar_vocales('murcielago') == 5
assert contar_vocales('xyz') == 0
print("✅ Ejercicio 4: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 4 ===
assert contar_vocales('') == 0
assert contar_vocales('aeiou') == 5
assert contar_vocales('programacion') == 5
print("✅ Ejercicio 4: tests adicionales superados.")
```

## Ejercicio 5 · Segundo número más grande (dos campeones)

Implementa `segundo_mayor(lista)` que devuelva el **segundo** valor más grande de una lista con al menos 2 elementos. Si el máximo se repite, el segundo mayor puede ser igual al mayor.

**Casos de ejemplo:** `segundo_mayor([3, 9, 7, 9])` → `9`; `segundo_mayor([1, 2])` → `1`.

FUNCIÓN segundo_mayor(lista):

    mayor = el más grande entre lista[0] y lista[1]
    segundo = el más pequeño entre lista[0] y lista[1]

    PARA cada candidato EN lista, empezando desde el índice 2:
        SI candidato > mayor:
            segundo = mayor        
            mayor = candidato        
        SINO SI candidato > segundo:
            segundo = candidato      

    RETORNAR segundo

```python
def segundo_mayor(lista):
    mayor = max(lista[0], lista[1])
    segundo = min(lista[0], lista[1])

    for c in lista[2:]:
        if c > mayor:
            segundo = mayor
            mayor = c
        elif c > segundo:
            segundo = c

    return segundo
```

```python
# === Tests visibles · Ejercicio 5 ===
assert segundo_mayor([3, 9, 7, 9]) == 9
assert segundo_mayor([1, 2]) == 1
print("✅ Ejercicio 5: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 5 ===
assert segundo_mayor([5, 4, 3, 2, 1]) == 4
assert segundo_mayor([-1, -2, -3]) == -2
assert segundo_mayor([10, 10, 10]) == 10
print("✅ Ejercicio 5: tests adicionales superados.")
```

## Ejercicio 6 · Invertir una lista (sin reverse ni [::-1])

Implementa `invertir(lista)` que devuelva una **nueva** lista con los elementos en orden inverso, **sin** usar `lista[::-1]` ni `.reverse()`.

**Casos de ejemplo:** `invertir([1, 2, 3])` → `[3, 2, 1]`; `invertir([])` → `[]`.

FUNCIÓN invertir(lista):

    nueva_lista = []

    PARA cada elemento EN lista (en orden original):
        nueva_lista = [elemento] + nueva_lista

    RETORNAR nueva_lista

```python
def invertir(lista):
    nueva_lista = []
    for e in lista:
        nueva_lista = [e] + nueva_lista
    return nueva_lista
```

```python
# === Tests visibles · Ejercicio 6 ===
assert invertir([1, 2, 3]) == [3, 2, 1]
assert invertir([]) == []
print("✅ Ejercicio 6: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 6 ===
assert invertir([7]) == [7]
assert invertir(['a', 'b']) == ['b', 'a']
assert invertir([1, 1, 2]) == [2, 1, 1]
print("✅ Ejercicio 6: tests adicionales superados.")
```

## Ejercicio 7 · ¿Es primo? (bucle + salida temprana)

Implementa `es_primo(n)` que devuelva `True` si `n` es primo y `False` si no. Recuerda: los primos son enteros `>= 2` divisibles solo por 1 y por sí mismos. `0` y `1` no son primos.

**Casos de ejemplo:** `es_primo(7)` → `True`; `es_primo(1)` → `False`; `es_primo(9)` → `False`.

FUNCIÓN es_primo(n):

    SI n < 2:
        RETORNAR False        

    PARA cada divisor desde 2 hasta n-1:
        SI n es divisible entre divisor (n % divisor == 0):
            RETORNAR False      

    RETORNAR True

```python
def es_primo(n):
  if n <= 1:
    return False
  for divisor in range(2, n):
    if n % divisor == 0:
      return False
  return True
```

```python
# === Tests visibles · Ejercicio 7 ===
assert es_primo(7) is True
assert es_primo(1) is False
assert es_primo(9) is False
print("✅ Ejercicio 7: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 7 ===
assert es_primo(2) is True
assert es_primo(0) is False
assert es_primo(97) is True
assert es_primo(100) is False
print("✅ Ejercicio 7: tests adicionales superados.")
```

## Ejercicio 8 · La moda: el valor más frecuente (contador)

Implementa `moda(lista)` que devuelva el elemento que **más veces** aparece en una lista no vacía. Si hay empate, devuelve el que aparece primero al recorrer la lista.

**Casos de ejemplo:** `moda([1, 2, 2, 3])` → `2`; `moda([5])` → `5`.

FUNCIÓN moda(lista):

    campeon = lista[0]
    mejor_conteo = número de veces que aparece lista[0] en la lista

    PARA cada numero EN lista:
        conteo_actual = número de veces que aparece numero en la lista

        SI conteo_actual > mejor_conteo:
            campeon = numero
            mejor_conteo = conteo_actual

    RETORNAR campeon

```python
def moda(lista):
    campeon = lista[0]
    mejor_conteo = lista.count(lista[0])

    for n in lista:
        conteo_actual = lista.count(n)
        if conteo_actual > mejor_conteo:
            campeon = n
            mejor_conteo = conteo_actual

    return campeon
```

```python
# === Tests visibles · Ejercicio 8 ===
assert moda([1, 2, 2, 3]) == 2
assert moda([5]) == 5
print("✅ Ejercicio 8: tests visibles superados.")
```

```python
# === Tests adicionales (ocultos) · Ejercicio 8 ===
assert moda([4, 4, 4, 1, 1]) == 4
assert moda(['a', 'b', 'a']) == 'a'
assert moda([7, 8, 8, 7, 7]) == 7
print("✅ Ejercicio 8: tests adicionales superados.")
```
